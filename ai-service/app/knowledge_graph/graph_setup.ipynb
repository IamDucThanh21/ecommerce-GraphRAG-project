```python
from neo4j import GraphDatabase

# Thông tin kết nối lấy từ Neo4j Desktop
URI = "neo4j://127.0.0.1:7687"
AUTH = ("neo4j", "12345678") # Thay password của bạn vào đây

def run_query(query, parameters=None):
    # Thêm cấu hình routing=False nếu cần, nhưng driver mặc định dựa vào URI
    with GraphDatabase.driver(URI, auth=AUTH) as driver:
        try:
            driver.verify_connectivity()
            with driver.session() as session:
                result = session.run(query, parameters)
                return [record for record in result]
        except Exception as e:
            print(f"Lỗi kết nối: {e}")

print("Kết nối thành công!")
```

```python
create_schema_query = """
// 1. Tạo Constraints để tránh trùng lặp
CREATE CONSTRAINT IF NOT EXISTS FOR (s:Series) REQUIRE s.name IS UNIQUE;
CREATE CONSTRAINT IF NOT EXISTS FOR (m:Model) REQUIRE m.name IS UNIQUE;

// 2. Tạo Series và Models
MERGE (series:Series {name: 'iPhone 17 Series'})
MERGE (m1:Model {name: 'iPhone 17 Pro'})
MERGE (m2:Model {name: 'iPhone 17'})
MERGE (m1)-[:BELONGS_TO]->(series)
MERGE (m2)-[:BELONGS_TO]->(series)

// 3. Tạo các thuộc tính dùng chung
MERGE (c1:Color {name: 'Cam Vũ Trụ'})
MERGE (c2:Color {name: 'Tím Oải Hương'})
MERGE (st:Storage {size: '256GB'})

// 4. Tạo ProductItem (SKU) và kết nối
MERGE (sku:ProductItem {sku_id: 'ip17p-256-orange', name: 'iPhone 17 Pro 256GB Cam'})
MERGE (sku)-[:IS_MODEL]->(m1)
MERGE (sku)-[:HAS_COLOR]->(c1)
MERGE (sku)-[:HAS_STORAGE]->(st)
"""

run_query(create_schema_query)
print("Đã khởi tạo xong cấu trúc Graph cơ bản.")
```
